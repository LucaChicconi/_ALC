## Ejercicio 4

Encontrar los coeficientes de la parábola y = ax^2 + bx + c que pasa por los puntos (1,1), (2,2) y (3,0). Verificar el resultado obtenido usando Python. Graficar los puntos y la parábola aprovechando el siguiente código:

````python
import matplotlib.pyplot as plt
# ...
# Acá, crear la matriz y resolver el sistema para calcular a,b y c
# ...
A = np.array([[1,1,1],[4,2,1],[9,3,1]])
v = np.array([1,2,0])

s = np.linalg.solve(A,v)

print(s[0])
print(s[1])
print(s[2])
xx = np.array([1,2,3])
yy = np.array([1,2,0])

x = np.linspace(0,4,100) #Genera 100 puntos equiespaciados entre 0 y 4
f = lambda t: s[0]*t**2+s[1]*t+s[2] #ésto genera una función f(t)

plt.plot(xx,yy,'*')
plt.plot(x,f(x))
plt.show
````
