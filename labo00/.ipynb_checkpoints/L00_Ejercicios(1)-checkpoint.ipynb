# **Ejercicio 4.**
Desarrollar una función `diagonal(A)` que devuelva la matriz $D$ correspondiente a la matriz diagonal de $A$.


```python
# Completa el código del Ejercicio 4 aquí
def diagonal(A):
    D = [[]]
    if len(A) != len(A[0]):
        return "La matriz no es cuadrada"
    else:
        D =[[val if i == j else 0 for j,val in enumerate(fila)]for i,fila in enumerate(A)]

    return np.array(D)
```

# **Ejercicio 5.**
Desarrollar una función `traza(A)` que calcule la traza de una matriz cualquiera $A$.


```python
# Completa el código del Ejercicio 5 aquí
def traza(A):
    sumaDiag=0
    filas = len(A)
    columnas = len(A[0])
    if filas != columnas:
        return "La matriz no es cuadrada!"
    else:
        for i in range(len(A)):
            sumaDiag += A[i][i]
    return sumaDiag
```

# **Ejercicio 6.**
Desarrollar una función `traspuesta(A)` que devuelva la matriz traspuesta de $A$.


```python
# Completa el código del Ejercicio 6 aquí
def traspuesta(A):
    A_arr = np.array(A)
    if A_arr.size()==0
        return A_arr
    if len(A_arr.shape) == 1:
        return A_arr
    T = [[fila[i] for fila in A] for i in range(len(A[0]))]
    
    return np.array(T)

```
