# **Ejercicio 8.**
Desarrollar una función `calcularAx(A, x)` que recibe una matriz $A$ de tamaño $n \times m$ y un vector $x$ de largo $m$ y devuelve un vector $b$ de largo $n$ resultado de la multiplicación vectorial de la matriz y el vector.


```python
# Completa el código del Ejercicio 8 aquí
def calcularAx(A,x):
    if len(A) == 0:
        return "No es posible hacer Ax. La matriz está vacía"
    filas = len(A)
    columnas = len(A[0])
    res = []
    
    if columnas == 0 or (columnas != len(x)):
        return "No es posible hacer Ax. No coinciden las dimensiones"
    else:
        for fila in range(filas):
            suma = 0
            for columna in range(len(x)):
                suma += A[fila][columna] * x[columna]
            res.append(suma)
        return res
            
        
```

# **Ejercicio 10.**
Desarrollar una función `sumar_fila_multiplo(A, i, j, s)` que a la fila $i$ le sume la fila $j$ multiplicada por un escalar $s$. Esta es una operación elemental clave en la eliminación gaussiana. La operación debe ser *in-place*.


```python
# Completa el código del Ejercicio 10 aquí
def sumar_fila_multiplo(A, i, j, s):
    if 0 <= i-1 < len(A) and 0 <= j-1 < len(A):
        fila_origen = list(A[j-1])

        for k in range(len(A[0])):
            A[i-1][k] += s * fila_origen[k]
    return np.array(A)
```

# **Ejercicio 11.** 
Desarrollar una función `esDiagonalmenteDominante(A)` que devuelva `True` si una matriz cuadrada $A$ es estrictamente diagonalmente dominante. Esto ocurre si para cada fila, el valor absoluto del elemento en la diagonal es mayor que la suma de los valores absolutos de los demás elementos en esa fila.


```python
# Completa el código del Ejercicio 11 aquí
def esDiagonalmenteDominante(A):
    if len(A) == 0:
        return None
        
    filas = len(A)
    columnas = len(A[0])
    
    if filas != columnas:
        return None
    else:
        for i in range(filas): # n veces O(n^2)
            suma = 0
            for columna in range(columnas):
                suma += A[i][columna]
            suma_sin_diagonal = suma - abs(A[i][i])
            if(abs(A[i][i]) <= suma_sin_diagonal):
                return False
        return True
```

# **Ejercicio 12.** 
Desarrollar una función `matrizCirculante(v)` que genere una matriz circulante a partir de un vector. En una matriz circulante la primer fila es igual al vector $v$, y en cada fila se encuentra una permutación cíclica de la fila anterior, moviendo los elementos un lugar hacia la derecha.


```python
# Completa el código del Ejercicio 12 aquí
def matrizCirculante(v):
    if len(v) == 0:
        return "Error: Vector vacío"
    tamaño = len(v)
    matriz = []
    for fila in range(tamaño):
        fila_nueva= []
        for columna in range(tamaño):
            fila_nueva.append(v[columna-fila % tamaño])
        matriz.append(fila_nueva)
    return np.array(matriz)
            
```

```python
# Completa el código del Ejercicio 13 aquí
def matrizVandermonde(v):
    if len(v)==0:
        return "Error: vector vacío"
    tamaño = len(v)
    matrizV=[]

    for fila in range(tamaño):
        fila_nueva=[]
        for columna in range(tamaño):
            fila_nueva.append(v[columna]**fila)
        matrizV.append(fila_nueva)
    return np.array(matrizV)
    
```

# **Ejercicio 14.**
Desarrollar una función `numeroAureo(n)` que estime el número áureo $\phi$ como $F_{k+1}/F_k$ siendo $F_k$ el $k$-ésimo número de la sucesión de Fibonacci. Para esto, formulen la sucesión de Fibonacci:
$$F_{k+1} = F_k + F_{k-1}$$
de forma matricial, usando la semilla $F_0 = 0$, $F_1 = 1$. Grafique el valor aproximado de $\phi$ en función del número de pasos de la sucesión considerado.


```python
import matplotlib.pyplot as plt
```

```python
# Completa el código del Ejercicio 14 aquí
matrizFib = [[1,1],
             [1,0]]

def multiplicar_matrices(A, B):
    C = [[0, 0], [0, 0]]
    for i in range(2):
        for j in range(2):
            for k in range(2):
                C[i][j] += A[i][k] * B[k][j]
    return C

def potencia_matriz(matriz, exponente):
    resultado = [[1, 0], [0, 1]]
    base = matriz
    
    for _ in range(exponente):
        resultado = multiplicar_matrices(resultado, base)
    
    return resultado

def numeroAureo(n):
    if n < 1:
        return "Error, división por 0"
    M_n = potencia_matriz(matrizFib,n)

    phi_k = M_n[0][0]/M_n[0][1]

    return phi_k

pasos = list(range(1, 11))
estimaciones = [numeroAureo(k) for k in pasos]

plt.figure(figsize=(8, 5))
plt.plot(pasos, estimaciones, marker='o', color='blue', label='Estimación de $\phi$')
plt.axhline(y=1.6180339887, color='red', linestyle='--', label='Valor real $\phi$ (~1.61803)')

plt.title('Aproximación del Número Áureo mediante Fibonacci Matricial')
plt.xlabel('Número de pasos (k)')
plt.ylabel('Valor aproximado')
plt.xticks(pasos)
plt.grid(True)
plt.legend()
plt.show()

    
```

# **Ejercicio 15.**
Desarrollar una función `matrizFibonacci(n)`, que genera una matriz $A$ de $n \times n$, y cada $a_{ij} = F_{i+j}$, siendo $F_k$ el $k$-ésimo número de la sucesión de Fibonacci (considerando $F_0 = 0, F_1 = 1$).


```python
# Completa el código del Ejercicio 15 aquí
m_fib = [[1,1],
         [1,0]]

def matrizFibonacci(n):
    if n <= 0:
        return "Error: tamaño de matriz inválido"
    if n == 1:
        return np.array(m_fib[0][0])
    max_index = 2 * n -2

    fib = [0, 1]
    while len(fib) <= max_index:
        fib.append(fib[-1] + fib[-2])

    matriz = []
    for i in range(n):
        fila = [fib[i + j] for j in range(n)]
        matriz.append(fila)

    return np.array(matriz)
        
    
```
