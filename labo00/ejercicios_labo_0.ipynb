Desarrollar una función triangSup(A) que devuelva la matriz U correspondiente a la matriz Triangular Superior de A sin su diagonal.

```python

def triangSup(A):
    U = [[val if i<j else 0 for j, val in enumerate(fila)] for i, fila in enumerate(A)]
    return np.array(U)

M3 = np.array([[1,2,3],
               [1,-2,3],
               [1,2,-3]])

M4 = np.array([[1,2,3,4],
               [1,-2,3,5],
               [1,2,-3,7]])
print(triangSup(M3))
print(triangSup(M4))
```

#### Ejercicio 5
Desarrollar una función traza(A) que calcule la traza de una matriz cualquiera A

```python
# La traza de una matriz es la suma de los elementos de su diaognal

def traza(A):
    sumaDiag=0
    if not esCuadrada(A):
        print("La matriz no es cuadrada!")
    else:
        matDiag = diagonal(A)
        for i in range(len(matDiag)):
            sumaDiag += matDiag[i][i]
    return sumaDiag

print(traza(M3))
print(traza(M4))
```

#### Ejercicio 7
Desarrollar una función esSimetrica(A) que devuelve True si la matriz A es simétrica y False en caso contrario.

```python
def esSimetrica(A):
    if len(A)!= len(A[0]):
        return False
    else:
        for i in range(len(A)):
            for j in range(i): # No hace falta mirar TODAS las filas y columnas
                if A[i][j] != A[j][i]:
                    return False
    return True

print(esSimetrica(M3))
print(esSimetrica(M4))

S1 = np.array([[1,2,3],
               [2,1,4],
               [3,4,1]])

print(esSimetrica(S1))
```

Desarrollar una función calcularAx(A,x) que recibe una matriz A de tamaño nxm y un vector x de largo m y devuelve un vector b de largo n resultado de la multiplicación vectorial de la matriz y el vector

```python
# Desarrollar una función calcularAx(A,x) que recibe una matriz A de tamaño nxm y un vector x de largo m y devuelve un vector b de largo n resultado de la multiplicación vectorial de la matriz y el vector

def calcularAx(A,x):
    if len(A[0]) != len(x):
        return None
    else:
        b= [0]*len(A)  # Hago un vector placeholder de longitud "filas(A)"
        for i in range(len(A)):
            for j in range(len(x)):
                b[i] += A[i][j] * x[j]
        return np.array(b)

#np.array(calcularAx(M3, np.array([1,2,3]))).shape

calcularAx(M3, np.array([1,2,3]))
```

#### Ejercicio 9
Desarrollar una función sumar_fila_multiplo(A,i,j,s) que a la fila i le sume la fila j multiplicada por un escalar s. Ésta es una operación clave en la eliminación gaussiana. La operación debe ser in-place 

```python
def sumar_fila_multiplo(A, i, j, s):
    if 0 <= i-1 < len(A) and 0 <= j-1 < len(A):
        fila_origen = list(A[j-1])

        for k in range(len(A[0])):
            A[i-1][k] += s * fila_origen[k]
    return A

M5 = [
    [2.0, 1.0, 1.0],
    [1.0, 2.0, 1.0],
    [1.0, 1.0, 2.0]
]

M5 = np.array(sumar_fila_multiplo(M5, 2, 1, -0.5))
sumar_fila_multiplo(M5, 3, 1, -0.5)
sumar_fila_multiplo(M5,3, 2, -(0.5/1.5))
print(M5)
```
