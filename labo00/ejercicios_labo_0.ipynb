#### Ejercicio 5
Desarrollar una función traza(A) que calcule la traza de una matriz cualquiera A

```python
# La traza de una matriz es la suma de los elementos de su diaognal

def traza(A):
    sumaDiag=0
    if not esCuadrada(A):
        print("La matriz no es cuadrada!")
    else:
        matDiag = diagonal(A)
        for i in range(len(matDiag)):
            sumaDiag += matDiag[i][i]
    return sumaDiag

print(traza(M3))
print(traza(M4))
```

```python
def esSimetrica(A):
    if len(A)!= len(A[0]):
        return False
    else:
        for i in range(len(A)):
            for j in range(i): # No hace falta mirar TODAS las filas y columnas
                if A[i][j] != A[j][i]:
                    return False
    return True

print(esSimetrica(M3))
print(esSimetrica(M4))

S1 = np.array([[1,2,3],
               [2,1,4],
               [3,4,1]])

print(esSimetrica(S1))
```

```python
# Desarrollar una función calcularAx(A,x) que recibe una matriz A de tamaño nxm y un vector x de largo m y devuelve un vector b de largo n resultado de la multiplicación vectorial de la matriz y el vector

def calcularAx(A,x):
    if len(A[0]) != len(x):
        return None
    else:
        b= [0]*len(A)  # Hago un vector placeholder de longitud "filas(A)"
        for i in range(len(A)):
            for j in range(len(x)):
                b[i] += A[i][j] * x[j]
        return np.array(b)

#np.array(calcularAx(M3, np.array([1,2,3]))).shape

calcularAx(M3, np.array([1,2,3]))
```

#### Ejercicio 11
Desarrollar una función esDiagonalmenteDominante(A) que devuelva True si una matriz cuadrada A es estrictamente diagonalmente dominante. Eso ocurre si **para cada fila, el valor absoluto del elemento en la diagonal es mayor que la suma de los valores absolutos de los demás elementos en esa fila**

```python
def esDiagonalmenteDominante(A):
    
    if len(A) != len(A[0]):
        return None
    else:
        for i in range(len(A)): # n veces O(n^2)
            suma_sin_diagonal = sum(abs(A[i])) - abs(A[i][i])
            if(abs(A[i][i]) <= suma_sin_diagonal):
                return False
        return True

matrix = np.array([[90,2,3],[3,50,7],[2,2,10]])

esDiagonalmenteDominante(matrix)
```

```python
matrix2 = np.array([
    [5, 1, 1],
    [9, 2, 9],
    [1, 2, 1]
])
esDiagonalmenteDominante(matrix2)

```

#### Ejercicio 12
Desarrollar una función matrizCirculante(v) que genere una matriz circulante a partir de un vector. En una matriz circulante, **la primer fila es igual al vector v y en cada fila se encuentra una permutación cíclica de la fila anterior moviendo los elementos un lugar hacia la derecha**

```python
def matrizCirculante(v):
    tamaño= len(v)
    return np.array([[v[(j-i) % tamaño]for j in range(tamaño)] for i in range(tamaño)])

matrizCirculante([1,2,3])
```
