# Ejercicio 1
La función float() genera un número flotante a partir de un número o un string.

Cuando se usa la función print() se corrobora que el número 0.1 es representado como 0.1 exacto. Internamente Python sólo puede representar 0.1 de forma aprox. Para poder ver la representación completa tenemos que usar la función format()

El código siguiente permite apreciar como Python representa 0.1 con 20 dígitos:

```py
format(0.1,'.20f')
```

Ésto nos lleva a 
```py
x = 0.1 + 0.1+ 0.1
y = 0.3

print(x==y)
```

Que tiene salida
```py
False
```

Utilizar format para verificar si la respuesta es correcta

````python
x = 0.1 + 0.1+ 0.1
y = 0.3
print(format(x,'.20f'))
print(format(y,'.20f'))

print(x==y)
````

# Ejercicio 2
Identificar la fuente del error y proponer una forma de solucionarlo

````python
a =  1.0
while a != format(0.1,'.1f'):
    print(a)
    a = float(a)
    a = format(a - 0.1,'.1f')
print('fin')
````
