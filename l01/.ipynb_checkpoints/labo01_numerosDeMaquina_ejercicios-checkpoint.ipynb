# **Ejercicio 1 (Representación de Flotantes).** 
Los números flotantes se representan como

La función `float()` genera un número flotante a partir de un número o de un string.
```python
>>> float(0.1)
0.1
>>> float('1.25')
1.25
```
Cuando se usa la función `print()`, se corrobora que el número 0.1 es representado como 0.1 exacto. Internamente Python solo puede representar 0.1 de forma aproximada. Para poder ver como la representación completa, se puede usar la función `format()`.
El código siguiente permite apreciar como Python es capaz de representar 0.1 usando 20 dígitos:
```python
>>> format(0.1, '.20f')
'0.10000000000000000555'
```
Esto nos lleva a
```python
>>> x = 0.1 + 0.1 + 0.1
>>> y = 0.3
>>> print(x == y)
```
y con salida
`False`

Utilizar `format()` para verificar si la respuesta es correcta.


````python
# Completa el código del Ejercicio 1 aquí
x = 0.1 + 0.1+ 0.1
y = 0.3
print(format(x,'.20f'))
print(format(y,'.20f'))

print(x==y)
````

````python
#2 y 3

def base2(numero):
    exponente = 0
    numDecimal = numero
# Normalizamos para que la mantisa quede en el rango [1.0, 2.0)
    if numero > 0:
        while numero < 1.0:
            numero *= 2
            exponente -= 1
        while numero >= 2.0:
            numero /= 2
            exponente += 1

    mantisa = numero
    return(f"Numero en decimal:{numDecimal} \n Mantisa: {mantisa} \n Exponente:{exponente}")


print(base2(0.3))
````

# **Ejercicio 7 (Arrastre de error: Descomposición LU).**
Desarrollar una función `matricesIguales(A, B)` que devuelve `True` si ambas matrices son iguales.

La siguiente matriz $A$:
$A = \begin{pmatrix} 4. & 2. & 1. \\ 2. & 7. & 9. \\ 0. & 5. & \frac{22}{3} \end{pmatrix}$

tiene una descomposición LU siguiente:
$L = \begin{pmatrix} 1. & 0. & 0. \\ 0.5 & 1. & 0. \\ 0. & \frac{5}{6} & 1. \end{pmatrix}$, $U = \begin{pmatrix} 4. & 2. & 1. \\ 0. & 6. & 8.5 \\ 0. & 0. & 0.25 \end{pmatrix}$

Verificar que la función desarrollada `matricesIguales(A, L @ U)` devuelva `True`.


````python
# Completa el código del Ejercicio 7 aquí
def matricesIguales(A,B):
    if A.shape != B.shape:
        return False
    for i in range(len(A)):
        for j in range(len(A[0])):
            if A[i][j] != B[i][j]:
                return False
    return True

A = np.array([
             [4.,2.,1.],
             [2.,7.,9.],
             [0.,5.,22/3]
])

L = np.array([[1.,0.,0.],
              [0.5,1.,0.],
              [0.,5/6,1.]])

U = np.array([[4.,2.,1.],
              [0.,6.,8.5],
              [0.,0.,0.25]])


print(L@U)
print(A)

matricesIguales(A, L@U)
````

# **Ejercicio 8 (Arrastre de error: función esSimetrica).**
Verificar que devuelve la función `esSimetrica()` desarrollada para el laboratorio pasado en los siguientes casos. Para una `A = np.array(np.random.rand(4,4))`
* `esSimetrica(A.T @ A)`
* `esSimetrica(A.T @ (A * 0.25) / 0.25)`
* `esSimetrica(A.T @ (A * 0.2) / 0.2)`


````python
def esSimetrica(A):
    if len(A)!= len(A[0]):
        return False
    else:
        for i in range(len(A)):
            for j in range(i): # No hace falta mirar TODAS las filas y columnas
                if A[i][j] != A[j][i]:
                    return False
    return True
````

````python
# Completa el código del Ejercicio 8 aquí
A = np.array(np.random.rand(4,4))

print(f"A.T @ A : {esSimetrica(A.T @ A)}")
print(f"A.T @ A : {esSimetrica(A.T @ (A*0.25)/0.25)}")
print(f"A.T @ A : {esSimetrica(A.T @ (A*0.2)/0.2)}")
````

````python
def matricesIguales(A, B, error_tolerado=1e-8):
    if not A or not B:
        return len(A) == len(B)
    if len(A) != len(B):
        return False

    filas = len(A)
    
    for i in range(filas):
        if len(A[i]) != len(B[i]):
            return False
            
        for j in range(len(A[i])):
            if abs(A[i][j] - B[i][j]) > error_tolerado:
                return False

    return True
````
